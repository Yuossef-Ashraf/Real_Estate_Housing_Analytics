```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
```

```python
file = pd.read_csv("/content/sample_data/mission 2.csv")
```

```python
file.head()
```

```python
file['Location'].value_counts()
```

```python
file.info()
```

```python
#filling na null values
file['Bedrooms'].fillna(file['Bedrooms'].median(), inplace=True)
```

```python
file.info()
```

```python
file.dropna(inplace=True)
#in case you want to delete null values
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
file['Location'] = le.fit_transform(file['Location'])
```

```python
file.head()
```

```python
# divide date
#x ==> features
#y  ==> target
```

```python
x = file.drop('Price', axis=1)
x
```

```python
y = file['Price']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
dt = DecisionTreeRegressor()
dt.fit(x_train, y_train)
```

```python
y_pred = dt.predict(x_test)
```

```python
#accuracy ==> classifciation
#r2_score ==> regreesion
from sklearn.metrics import r2_score
print(r2_score(y_pred, y_test) * 100, '%')
```
