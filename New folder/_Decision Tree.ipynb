```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt

# Step 1: Simple dataset - "Should I buy an umbrella?"
data = {
    'Rainy': ['Yes', 'Yes', 'No', 'No'],
    'Windy': ['Yes', 'No', 'Yes', 'No'],
    'BuyUmbrella': ['No', 'Yes', 'No', 'No']
}

df = pd.DataFrame(data)

# Step 2: Encode categorical variables
df['Rainy'] = df['Rainy'].map({'Yes': 1, 'No': 0})
df['Windy'] = df['Windy'].map({'Yes': 1, 'No': 0})
df['BuyUmbrella'] = df['BuyUmbrella'].map({'Yes': 1, 'No': 0})

# Step 3: Features and target
X = df[['Rainy', 'Windy']]
y = df['BuyUmbrella']

# Step 4: Train decision tree model
model = DecisionTreeClassifier(criterion='entropy', max_depth=3)  # add max_depth for better control
model.fit(X, y)

# Step 5: Plot the decision tree
plt.figure(figsize=(10, 6))
tree.plot_tree(model,
               feature_names=['Rainy', 'Windy'],
               class_names=['No', 'Yes'],
               filled=True,
               rounded=True,
               fontsize=12)
plt.title("Decision Tree: Should I Buy an Umbrella?")
plt.show()
```
